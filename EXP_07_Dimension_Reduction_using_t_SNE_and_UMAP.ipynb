# Step 1 --Import the required libraries

```python
# Experiment 7: Dimension Reduction using t-SNE and UMAP

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE

import umap.umap_ as umap

print("Libraries imported successfully.")
```

# Step 2 — Load the dataset

```python
# Load the Digits dataset

digits = load_digits()

X = digits.data
y = digits.target

print("Dataset shape:", X.shape)
print("Target shape:", y.shape)
print("Number of classes:", len(np.unique(y)))
print("Classes:", np.unique(y))
```

# Step 3 — Visualize some original images

```python
# Display sample images from the dataset

fig, axes = plt.subplots(2, 5, figsize=(10, 5))

for i, ax in enumerate(axes.ravel()):
    ax.imshow(digits.images[i], cmap="gray")
    ax.set_title(f"Digit: {y[i]}")
    ax.axis("off")

plt.suptitle("Sample Images from Digits Dataset")
plt.tight_layout()
plt.show()
```

# Step 4 — Standardize the data

```python
# Standardize the features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
print("Mean of first feature:", round(X_scaled[:, 0].mean(), 4))
print("Standard deviation of first feature:", round(X_scaled[:, 0].std(), 4))
```

# Step 5 — t-SNE for 2D

```python
# t-SNE dimensionality reduction to 2D

tsne_2d = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    init="pca"
)

X_tsne_2d = tsne_2d.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape[1])
print("Reduced dimensions:", X_tsne_2d.shape[1])
```

# Step 6 — Visualize t-SNE 2D

```python
# Plot t-SNE 2D result

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_tsne_2d[:, 0],
    X_tsne_2d[:, 1],
    c=y,
    cmap="tab10",
    s=15
)

plt.colorbar(scatter, label="Digit")
plt.title("t-SNE Dimension Reduction to 2D")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.show()
```

# Step 7 — t-SNE for 3D

```python
# t-SNE dimensionality reduction to 3D

tsne_3d = TSNE(
    n_components=3,
    random_state=42,
    perplexity=30,
    init="pca"
)

X_tsne_3d = tsne_3d.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape[1])
print("Reduced dimensions:", X_tsne_3d.shape[1])
```

# Step 8 — Visualize t-SNE 3D

```python
# Plot t-SNE 3D result

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    X_tsne_3d[:, 0],
    X_tsne_3d[:, 1],
    X_tsne_3d[:, 2],
    c=y,
    cmap="tab10",
    s=15
)

ax.set_title("t-SNE Dimension Reduction to 3D")
ax.set_xlabel("t-SNE Component 1")
ax.set_ylabel("t-SNE Component 2")
ax.set_zlabel("t-SNE Component 3")

fig.colorbar(scatter, ax=ax, label="Digit")
plt.show()
```

# Step 9 — UMAP for 2D

```python
# UMAP dimensionality reduction to 2D

umap_2d = umap.UMAP(
    n_components=2,
    random_state=42
)

X_umap_2d = umap_2d.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape[1])
print("Reduced dimensions:", X_umap_2d.shape[1])
```

# Step 10 — Visualize UMAP 2D

```python
# Plot UMAP 2D result

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_umap_2d[:, 0],
    X_umap_2d[:, 1],
    c=y,
    cmap="tab10",
    s=15
)

plt.colorbar(scatter, label="Digit")
plt.title("UMAP Dimension Reduction to 2D")
plt.xlabel("UMAP Component 1")
plt.ylabel("UMAP Component 2")
plt.show()
```

# Step 11 — UMAP for 3D

```python
# UMAP dimensionality reduction to 3D

umap_3d = umap.UMAP(
    n_components=3,
    random_state=42
)

X_umap_3d = umap_3d.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape[1])
print("Reduced dimensions:", X_umap_3d.shape[1])
```

# Step 12 — Visualize UMAP 3D

```python
# Plot UMAP 3D result

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    X_umap_3d[:, 0],
    X_umap_3d[:, 1],
    X_umap_3d[:, 2],
    c=y,
    cmap="tab10",
    s=15
)

ax.set_title("UMAP Dimension Reduction to 3D")
ax.set_xlabel("UMAP Component 1")
ax.set_ylabel("UMAP Component 2")
ax.set_zlabel("UMAP Component 3")

fig.colorbar(scatter, ax=ax, label="Digit")
plt.show()
```

# Step 13 — Final comparison

```python
# Compare t-SNE and UMAP 2D results

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].scatter(
    X_tsne_2d[:, 0],
    X_tsne_2d[:, 1],
    c=y,
    cmap="tab10",
    s=15
)
axes[0].set_title("t-SNE 2D")
axes[0].set_xlabel("Component 1")
axes[0].set_ylabel("Component 2")

axes[1].scatter(
    X_umap_2d[:, 0],
    X_umap_2d[:, 1],
    c=y,
    cmap="tab10",
    s=15
)
axes[1].set_title("UMAP 2D")
axes[1].set_xlabel("Component 1")
axes[1].set_ylabel("Component 2")

plt.suptitle("Comparison of t-SNE and UMAP")
plt.tight_layout()
plt.show()
```

# Conclusion:
Dimensionality reduction was successfully performed on the Digits dataset
using t-SNE and UMAP. The original 64-dimensional feature space was reduced
to both 2D and 3D representations. The resulting visualizations show that
both techniques preserve meaningful neighborhood and class structure, making
it easier to visualize the distribution of the handwritten digit classes.
